# 01 — Exploratory Data Analysis & Risk Profile Assessment
**Dataset:** Home Credit Default Risk (`application_train.csv`)
**Objective:** Analyze applicant demographics, financial distributions, missingness patterns, and default risk signals.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

raw_path = '../data/raw/application_train.csv'
df = pd.read_csv(raw_path)
print(f'Raw dataset shape: {df.shape}')
print('Target Class Distribution:')
print(df['TARGET'].value_counts(normalize=True))
```

### Key Finding: Severe Class Imbalance
- ~92.3% Non-Defaulters (`TARGET = 0`)
- ~7.7% Defaulters (`TARGET = 1`)
- 12:1 negative-to-positive ratio requires cost-sensitive weighting and PR-AUC/ROC-AUC evaluation rather than raw accuracy.

```python
# Missing values analysis
missing = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
print('Top 10 columns by missingness (%):')
print(missing.head(10))

# DAYS_EMPLOYED anomaly check
anom_count = (df['DAYS_EMPLOYED'] == 365243).sum()
print(f'DAYS_EMPLOYED anomaly count (365243): {anom_count} ({anom_count/len(df):.1%})')
```
