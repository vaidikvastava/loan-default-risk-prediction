# 02 — Financial Feature Engineering & Ratio Derivations
**Objective:** Construct domain-specific affordability metrics, external bureau aggregates, and robust categorical encodings.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/application_train.csv')

# Domain Ratios
df['DEBT_TO_INCOME'] = df['AMT_CREDIT'] / df['AMT_INCOME_TOTAL']
df['ANNUITY_TO_INCOME'] = df['AMT_ANNUITY'] / df['AMT_INCOME_TOTAL']
df['CREDIT_TO_GOODS_RATIO'] = df['AMT_CREDIT'] / df['AMT_GOODS_PRICE']
df['CREDIT_ANNUITY_RATIO'] = df['AMT_CREDIT'] / df['AMT_ANNUITY']

# External ratings
ext_cols = ['EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3']
df['EXT_SOURCE_MEAN'] = df[ext_cols].mean(axis=1)
print('Engineered financial ratios preview:')
print(df[['DEBT_TO_INCOME', 'ANNUITY_TO_INCOME', 'EXT_SOURCE_MEAN']].describe())
```
